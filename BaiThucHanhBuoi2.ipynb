```
import pandas as pd
import matplotlib.pyplot as plt

# Đọc dữ liệu
df = pd.read_csv("dulieuxettuyendaihoc.csv")

# Thống kê số lượng theo KT và KV
bang = df.groupby(['KT', 'KV']).size().unstack()

# In bảng
print(bang)

# Vẽ biểu đồ
bang.plot(kind='bar')

plt.title("Số lượng thí sinh theo KV và KT")

plt.xlabel("Khối thi")
plt.ylabel("Số lượng")

plt.legend(title='Khu vực')

plt.show()
```

```
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("dulieuxettuyendaihoc.csv")
plt.plot(df['T1'])

plt.title("Simple Line T1")
plt.show()
def xep_loai(diem):
    if diem < 5:
        return 'k'
    elif diem < 7:
        return 'tb'
    elif diem < 8:
        return 'kha'
    else:
        return 'g'

df['phanloaiT1'] = df['T1'].apply(xep_loai)
tan_so = df['phanloaiT1'].value_counts()

print(tan_so)
bang = df.groupby(['T1', 'phanloaiT1']).size().unstack(fill_value=0)

bang.plot()

plt.title("Multiple Line")
plt.show()
plt.stem(df.index,
         df['T1'])

plt.title("Drop-line T1")

plt.show()
```

```
def xep_loai(diem):
    if diem < 5:
        return 'k'
    elif diem < 7:
        return 'tb'
    elif diem < 8:
        return 'kha'
    else:
        return 'g'

df['phanloaiT1'] = df['T1'].apply(xep_loai)
```

```
import seaborn as sns

sns.boxplot(x='phanloaiT1',
            y='T1',
            data=df)

plt.title("Boxplot T1 theo phân loại")

plt.show()
```

```
for loai in df['phanloaiT1'].unique():

    data = df[df['phanloaiT1'] == loai]

    plt.hist(data['T1'],
             alpha=0.5,
             label=loai)

plt.legend()

plt.title("Histogram T1 theo phân loại")

plt.show()
```

```
for loai in df['phanloaiT1'].unique():

    data = df[df['phanloaiT1'] == loai]

    stats.probplot(data['T1'],
                   dist="norm",
                   plot=plt)

    plt.title(f"QQ-Plot nhóm {loai}")

    plt.show()
```
