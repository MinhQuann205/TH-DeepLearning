```
import pandas as pd

df = pd.read_csv("dulieuxettuyendaihoc.csv")

print(df.head())
```

```
sort_dh1 = df.sort_values(by='DH1', ascending=True)

print(sort_dh1[['STT', 'DH1']])
```

```
sort_dh2_gt = df.sort_values(by=['GT', 'DH2'],
                             ascending=[True, True])

print(sort_dh2_gt[['STT', 'GT', 'DH2']])
```

```
group = df.groupby('KT')['DH1']

pivot_kt = pd.DataFrame({
    'count': group.count(),
    'sum': group.sum(),
    'mean': group.mean(),
    'median': group.median(),
    'min': group.min(),
    'max': group.max(),
    'std': group.std(),
    'Q1': group.quantile(0.25),
    'Q2': group.quantile(0.50),
    'Q3': group.quantile(0.75)
})

print(pivot_kt)
```

```
group = df.groupby(['KT', 'KV'])['DH1']

pivot_kt_kv = pd.DataFrame({
    'count': group.count(),
    'sum': group.sum(),
    'mean': group.mean(),
    'median': group.median(),
    'min': group.min(),
    'max': group.max(),
    'std': group.std(),
    'Q1': group.quantile(0.25),
    'Q2': group.quantile(0.50),
    'Q3': group.quantile(0.75)
})

print(pivot_kt_kv)
```

```
group = df.groupby(['KT', 'KV', 'DT'])['DH1']

pivot_kt_kv_dt = pd.DataFrame({
    'count': group.count(),
    'sum': group.sum(),
    'mean': group.mean(),
    'median': group.median(),
    'min': group.min(),
    'max': group.max(),
    'std': group.std(),
    'Q1': group.quantile(0.25),
    'Q2': group.quantile(0.50),
    'Q3': group.quantile(0.75)
})

print(pivot_kt_kv_dt)
```

```
tan_so = df['GT'].value_counts()

print(tan_so)
```

```
tan_suat = df['GT'].value_counts(normalize=True)

print(tan_suat)
```

```
import matplotlib.pyplot as plt

tan_so.plot(kind='bar')

plt.title("Biểu đồ tần số GT")
plt.xlabel("Giới tính")
plt.ylabel("Số lượng")

plt.show()
```

```
tan_so.plot(kind='pie',
            autopct='%1.1f%%')

plt.title("Biểu đồ tần suất GT")

plt.ylabel("")
plt.show()
```

```
tich_luy = tan_suat.cumsum()

plt.plot(tich_luy.index,
         tich_luy.values,
         marker='o')

plt.title("Đa giác tích lũy tần suất GT")
plt.xlabel("GT")
plt.ylabel("Tích lũy")

plt.grid(True)

plt.show()
```

```
loc = df[
    (
        (df['DH1'] >= 5.0) &
        (df['DH2'] >= 4.0) &
        (df['DH3'] >= 4.0)
    )
]

print(loc[['KV', 'DT', 'DH1', 'DH2', 'DH3']])
```

```
thongke_kv = loc['KV'].value_counts()

print(thongke_kv)
```

```
thongke_kv.plot(kind='bar')

plt.title("Số học sinh Kinh đạt điều kiện theo KV")
plt.xlabel("KV")
plt.ylabel("Số lượng")

plt.show()
```

```
loc2 = df[
    (df['DH1'] >= 5) &
    (df['DH2'] >= 5) &
    (df['DH3'] >= 5) &
    (df['KV'] == '2NT')
]

print(loc2[['DH1', 'DH2', 'DH3', 'KV']])


plt.plot(loc2['DH1'].values,
         label='DH1')

plt.plot(loc2['DH2'].values,
         label='DH2')

plt.plot(loc2['DH3'].values,
         label='DH3')

plt.legend()

plt.title("Điểm DH1 DH2 DH3 của KV 2NT")

plt.show()
```

```
import pandas as pd
import matplotlib.pyplot as plt

# Đọc dữ liệu
df = pd.read_csv("dulieuxettuyendaihoc.csv")

# Thống kê số lượng theo KT và KV
bang = df.groupby(['KT', 'KV']).size().unstack()

# In bảng
print(bang)

# Vẽ biểu đồ
bang.plot(kind='bar')

plt.title("Số lượng thí sinh theo KV và KT")

plt.xlabel("Khối thi")
plt.ylabel("Số lượng")

plt.legend(title='Khu vực')

plt.show()
```
